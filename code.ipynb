# Categorical Values vs default

```python
categorical_cols = [
    "experience_c",
    "emp_length",
    "purpose",
    "home_ownership_n"
]
```

```python
for col in categorical_cols:

    default_rate = (
        df.groupby(col)["Default"]
          .mean()
          .mul(100)
          .sort_values(ascending=False)
    )

    plt.figure(figsize=(8, 4))

    sns.barplot(
        x=default_rate.index,
        y=default_rate.values
    )

    plt.title(f"Default Rate by {col}")
    plt.xlabel(col)
    plt.ylabel("Default Rate (%)")
    plt.xticks(rotation=45)

    plt.show()
```

# Feature Engineering

```python
X = df.drop(columns=["Default", "id", "issue_d"]).copy()

y = df["Default"].copy()

X["loan_to_income"] = (
    X["loan_amnt"] / X["revenue"].replace(0, np.nan)
)
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    stratify=y,
    random_state=42
)
```

```python
numeric_features = [
    "revenue",
    "dti_n",
    "loan_amnt",
    "fico_n",
    "experience_c",
    "loan_to_income"
]

categorical_features = [
    "emp_length",
    "purpose",
    "home_ownership_n"
]
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.linear_model import LogisticRegression
```

```python
numeric_transformer_lr = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])
```

```python
categorical_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(handle_unknown="ignore"))
])
```

```python
preprocessor_lr = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer_lr, numeric_features),
        ("cat", categorical_transformer, categorical_features)
    ]
)
```

```python
lr_pipeline = Pipeline(steps=[
    ("preprocessor", preprocessor_lr),
    ("model", LogisticRegression(
        max_iter=1000,
        random_state=42
    ))
])
```
